# 线性回归学习笔记

学习目标：理解损失函数、正规方程、梯度下降

核心代码：`LinearRegression/linear_regression.py`

```python
import sys
sys.path.insert(0, '..')
from LinearRegression.linear_regression import LinearRegression
import numpy as np

# 生成合成数据
np.random.seed(42)
X = np.random.randn(100, 1)
y = 3 * X.ravel() + 2 + np.random.randn(100) * 0.5
print('X shape:', X.shape, 'y shape:', y.shape)
```

```python
# 正规方程求解
model = LinearRegression()
model.fit(X, y, method='normal_eq')
print('weights:', model.weights)
print('bias:', model.bias)
print('R²:', model.score(X, y))
```

```python
# 梯度下降求解：改 lr 观察收敛
for lr in [0.001, 0.01, 0.1, 0.5]:
    model_gd = LinearRegression()
    model_gd.fit(X, y, method='gradient_descent', lr=lr, n_iters=1000)
    print(f'lr={lr}, R²={model_gd.score(X, y):.4f}')
```

## 实验

1. 改变 lr，观察收敛速度变化
2. 改变 n_iters，观察欠拟合/过拟合
3. 把 X 换成多维，看两种方法是否一致
