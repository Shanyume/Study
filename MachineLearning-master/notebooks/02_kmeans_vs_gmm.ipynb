# KMeans vs GMM 对比学习

硬聚类 vs 软聚类

- KMeans：每个样本只能属于一个簇
- GMM：每个样本属于每个簇都有概率

```python
import sys
sys.path.insert(0, '..')
from KMeans.kmeans import KMeans
from GMM.gmm import GMM, best_label_mapping
from sklearn.datasets import load_iris

X, y = load_iris(return_X_y=True)
```

```python
# KMeans
km = KMeans(n_clusters=3, max_iter=50)
km.fit(X)
print('KMeans SSE:', km.sse)
print('KMeans acc:', best_label_mapping(y, km.labels.astype(int), 3))
```

```python
# GMM
gmm = GMM(n_components=3)
gmm.fit(X)
print('GMM acc:', best_label_mapping(y, gmm.predict(X), 3))
```

## 观察

- GMM 在 Iris 上通常比 KMeans 准确率更高（软边界）
- 尝试改 n_clusters，看两种方法的变化
