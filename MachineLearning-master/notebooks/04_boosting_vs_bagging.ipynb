# Bagging vs Boosting 对比

两种集成学习策略的区别

```python
import sys
sys.path.insert(0, '..')
from Bagging.bagging import BaggingClassifier
from Boosting.GradientBoosting.gradient_boosting import GradientBoostingClassifier
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

X, y = make_moons(n_samples=500, noise=0.25, random_state=42)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Bagging：并行，降低方差
bag = BaggingClassifier(n_estimators=20)
bag.fit(Xtr, ytr)
print('Bagging test acc:', accuracy_score(yte, bag.predict(Xte)))
```

```python
# Gradient Boosting：串行，降低偏差
gb = GradientBoostingClassifier(n_estimators=100, learning_rate=0.2)
gb.fit(Xtr, ytr)
print('GB test acc:', accuracy_score(yte, gb.predict(Xte)))
```

## 核心区别

| | Bagging | Boosting |
|---|---|---|
| 训练方式 | 并行 | 串行 |
| 降低 | 方差 | 偏差 |
| 样本权重 | 相同 | 逐步调整（AdaBoost）或拟合残差（GB） |
| 代表 | Random Forest | XGBoost / LightGBM |
